```python
import numpy as np
import matplotlib.pyplot as plt

arr1 = np.array([[2,4],[6,3]])
arr2 = np.array([[5,7],[8,9]])
print(str(arr1+arr2)+"addition")
print("\n")
print(str(arr1-arr2)+"subtraction")
print("\n")
print(str(arr1*arr2)+"multiplication")
print("\n")
print(np.transpose(arr1))

```

```python
a= np.array([[4,11],[14,5]],dtype=float)
U,s,Vt=np.linalg.svd(a)
Sigma=np.diag(s)
print(np.round(U,4))
print(np.round(Sigma,4))
print(np.round(Vt,4))
```
