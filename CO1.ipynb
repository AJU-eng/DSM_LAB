```python
import matplotlib.pyplot as plt
x=[1,2,6,18]
y=[3,10,12,20]

plt.plot(x,y,linestyle=':',color='red',marker='o')
plt.title("Line graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
men = [22,30,35,35,26]
women = [25,32,30,35,29]
x = [1,2,3,4,5]
plt.bar(x,men,width=0.4,label="Men")
plt.bar([i+0.4 for i in x],women,width=0.4,label="Women")
plt.xlabel("Group")
plt.ylabel("Scores")
plt.legend()
plt.show()
```
